# campaign_history
* Jumlah paparan 20.000- 50.000
* respons 5000-10.000
* Semua customer id ada di CRM

```python
camp = pd.read_csv('campaign_history.csv')
camp
```

```python
import pandas as pd

# Buka file
camp = pd.read_csv('campaign_history.csv')

print("="*60)
print("📊 CEK JUMLAH PAPARAN, RESPON & KONVERSI")
print("="*60)

# 1. Jumlah Paparan = jumlah semua baris
jumlah_paparan = len(camp)
print(f"\n📢 1. JUMLAH PAPARAN KESELURUHAN: {jumlah_paparan:,}")
print(f"   Target dokumen: 20.000 – 50.000")
if 20000 <= jumlah_paparan <= 50000:
    print("   ✅ SESUAI TARGET")
else:
    print("   ⚠️ DI LUAR RENTANG TARGET")

# 2. Jumlah Respon = responded = True
jumlah_respon = camp['responded'].sum()
rasio_respon = (jumlah_respon / jumlah_paparan * 100) if jumlah_paparan > 0 else 0
print(f"\n✅ 2. JUMLAH RESPON: {jumlah_respon:,}")
print(f"   Target dokumen: 5.000 – 10.000")
if 5000 <= jumlah_respon <= 10000:
    print("   ✅ SESUAI TARGET")
else:
    print("   ⚠️ DI LUAR RENTANG TARGET")
print(f"   Tingkat respon: {rasio_respon:.1f}%")

# 3. Jumlah Konversi = converted = True
jumlah_konversi = camp['converted'].sum()
rasio_konversi = (jumlah_konversi / jumlah_paparan * 100) if jumlah_paparan > 0 else 0
print(f"\n💰 3. JUMLAH KONVERSI (berhasil belanja): {jumlah_konversi:,}")
print(f"   Rasio konversi: {rasio_konversi:.1f}%")

# 4. Per Kampanye
print("\n" + "="*60)
print("📈 RINCIAN PER KAMPANYE")
print("="*60)
ringkasan = camp.groupby('campaign_name').agg(
    jumlah_paparan=('responded', 'count'),
    jumlah_respon=('responded', 'sum'),
    jumlah_konversi=('converted', 'sum')
).reset_index()
ringkasan['rasio_respon_%'] = (ringkasan['jumlah_respon'] / ringkasan['jumlah_paparan'] * 100).round(1)
print(ringkasan.to_string(index=False))

# 5. Per Saluran (email/WA/push)
print("\n" + "="*60)
print("📱 RINCIAN PER SALURAN")
print("="*60)
saluran = camp.groupby('channel').agg(
    jumlah_paparan=('responded', 'count'),
    jumlah_respon=('responded', 'sum')
).reset_index()
saluran['rasio_respon_%'] = (saluran['jumlah_respon'] / saluran['jumlah_paparan'] * 100).round(1)
print(saluran.to_string(index=False))

print("\n" + "="*60)
print("✅ Selesai!")
print("="*60)
```

```python
import pandas as pd

# Buka semua file yang dibutuhkan
ident = pd.read_csv('customer_identities.csv')
trans = pd.read_csv('transactions.csv')
items = pd.read_csv('transaction_items.csv')

# Ambil contoh pelanggan yang mau dicek
cari_cid = 'CUST00003'

# 1. Ketahui semua source_id milik pelanggan ini
sumber = ident[ident['customer_id'] == cari_cid]
print(f"📋 Sistem yang dipakai {cari_cid}:")
print(sumber[['source_system', 'source_id']].to_string(index=False))

# 2. Cari SEMUA transaksi milik pelanggan ini
trans_cust = trans[trans['customer_id'] == cari_cid].copy()

# 3. Gabungkan dengan barang yang dibeli
trans_dengan_barang = trans_cust.merge(
    items, on='transaction_id', how='left'
)[['transaction_id', 'transaction_date', 'channel',
   'item_type', 'item_name', 'quantity', 'final_amount']]

print(f"\n🛒 Semua pembelian {cari_cid}:")
print(trans_dengan_barang.to_string(index=False))
print(f"\n💰 Total belanja: Rp{trans_cust['final_amount'].sum():,}")
```
